```python
import pandas as pd
import numpy as np

# Load original dataset
file_path = r'F:\Program Files (x86)\OneDrive\Desktop\Ml edit 17\Room Temp.xlsx'
df = pd.read_excel(file_path)

# Separate features and target
features = df.iloc[:, :-1]
target = df.iloc[:, -1]

# Define augmentation function
def augment_data(df_features, df_target, noise_level=0.03, times=2):
    augmented_features = []
    augmented_target = []

    for _ in range(times):
        # Gaussian noise scaled by feature values
        noise_feat = df_features * noise_level * np.random.randn(*df_features.shape)
        new_features = df_features + noise_feat

        # Ensure features stay positive
        new_features = new_features.clip(lower=1e-5)

        # Gaussian noise for target
        noise_targ = df_target * noise_level * np.random.randn(*df_target.shape)
        new_target = df_target + noise_targ

        # Ensure target stays positive
        new_target = new_target.clip(lower=1e-5)

        augmented_features.append(new_features)
        augmented_target.append(new_target)

    # Combine all augmented data
    features_aug = pd.concat([df_features] + augmented_features, ignore_index=True)
    target_aug = pd.concat([df_target] + augmented_target, ignore_index=True)

    # Final DataFrame
    df_aug = pd.concat([features_aug, target_aug], axis=1)
    df_aug.columns = df.columns  # Keep original column names

    return df_aug

# Run augmentation
augmented_df = augment_data(features, target, noise_level=0.03, times=2)

# Save the result
augmented_df.to_excel("Room_Temp___3x_Augmented_3Percent_Noise.xlsx", index=False)
print("✅ Augmentation complete. File saved: Room_Temp___3x_Augmented_3Percent_Noise.xlsx")
```
